```python
import torch
import platform
import os
import psutil
```

```python
print("========= SYSTEM INFO =========")
print(f"OS         : {platform.system()} {platform.release()}")
print(f"Processor  : {platform.processor()}")
print(f"CPU Cores  : {psutil.cpu_count(logical=True)}")
print(f"RAM        : {round(psutil.virtual_memory().total / (1024 ** 3), 2)} GB")
```

```python
print("\n========= PYTORCH + CUDA INFO =========")
print(f"PyTorch version : {torch.__version__}")
print(f"CUDA available  : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA version    : {torch.version.cuda}")
    print(f"cuDNN version   : {torch.backends.cudnn.version()}")
    print(f"GPU Name        : {torch.cuda.get_device_name(0)}")
    print(f"GPU Capabilities: {torch.cuda.get_device_capability(0)}")
    print(f"GPU Memory      : {round(torch.cuda.get_device_properties(0).total_memory / (1024 ** 3), 2)} GB")
    print(f"GPU Multiprocessors (SMs): {torch.cuda.get_device_properties(0).multi_processor_count}")
else:
    print("CUDA not available, GPU not being used.")
```

```python
print("\n========= PYTHON & LIBRARY VERSIONS =========")
print(f"Python version  : {platform.python_version()}")
try:
    import ultralytics
    print(f"Ultralytics YOLOv8 : {ultralytics.__version__}")
except:
    print("Ultralytics YOLOv8 not installed.")

try:
    import cv2
    print(f"OpenCV version     : {cv2.__version__}")
except:
    print("OpenCV not installed.")
```

```python
import torch
import platform
import os

# Basic system info
print(" Python version:", platform.python_version())
print(" CPU cores:", os.cpu_count())
print(" Total RAM (approx):", round(os.sysconf('SC_PAGE_SIZE') * os.sysconf('SC_PHYS_PAGES') / (1024 ** 3)), "GB")

# GPU info
if torch.cuda.is_available():
    print(" CUDA is available!")
    print(" Number of GPUs:", torch.cuda.device_count())
    print(" GPU Name:", torch.cuda.get_device_name(0))
    print(" CUDA version:", torch.version.cuda)
    print(" cuDNN version:", torch.backends.cudnn.version())
    print(" PyTorch version:", torch.__version__)
    print(" GPU memory total (GB):", round(torch.cuda.get_device_properties(0).total_memory / 1024 ** 3, 2))
else:
    print(" CUDA is NOT available.")

# YOLO version (if installed)
try:
    import ultralytics
    print(" Ultralytics YOLO version:", ultralytics.__version__)
except ImportError:
    print(" Ultralytics YOLO not installed.")
```
