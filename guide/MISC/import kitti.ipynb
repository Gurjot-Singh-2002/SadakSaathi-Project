```python
# Step 1: Download KITTI dataset files (training images + labels)
import os
import urllib.request
import zipfile

# Create a directory to store the data
os.makedirs("KITTI_raw", exist_ok=True)

# URLs from official KITTI site
image_url = "https://s3.eu-central-1.amazonaws.com/avg-kitti/data_object_image_2.zip"
label_url = "https://s3.eu-central-1.amazonaws.com/avg-kitti/data_object_label_2.zip"

image_zip_path = "KITTI_raw/data_object_image_2.zip"
label_zip_path = "KITTI_raw/data_object_label_2.zip"

# Download images
if not os.path.exists(image_zip_path):
    print("Downloading image dataset...")
    urllib.request.urlretrieve(image_url, image_zip_path)
    print("Image dataset downloaded!")

# Download labels
if not os.path.exists(label_zip_path):
    print("Downloading label dataset...")
    urllib.request.urlretrieve(label_url, label_zip_path)
    print("Label dataset downloaded!")
```

```python
# Step 2: Extract downloaded zip files
extract_dir = "KITTI"

with zipfile.ZipFile(image_zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

with zipfile.ZipFile(label_zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)
```

```python
# Step 3: Move folders to simplified structure
import shutil

# Create target folders
os.makedirs("KITTI/images", exist_ok=True)
os.makedirs("KITTI/labels", exist_ok=True)

# Move image files
src_images = os.path.join(extract_dir, "training", "image_2")
dst_images = os.path.join("KITTI", "images")
for file in os.listdir(src_images):
    shutil.move(os.path.join(src_images, file), dst_images)

# Move label files
src_labels = os.path.join(extract_dir, "training", "label_2")
dst_labels = os.path.join("KITTI", "labels")
for file in os.listdir(src_labels):
    shutil.move(os.path.join(src_labels, file), dst_labels)

# Remove old folders
shutil.rmtree(os.path.join(extract_dir, "training"))
```

```python
# Show one example image and corresponding label
from PIL import Image

sample_img = os.listdir("KITTI/images")[0]
sample_lbl = sample_img.replace('.png', '.txt')

img_path = os.path.join("KITTI/images", sample_img)
lbl_path = os.path.join("KITTI/labels", sample_lbl)

# Open and display the image
img = Image.open(img_path)
img.show()

# Print its labels
with open(lbl_path, 'r') as f:
    lines = f.readlines()
    for line in lines:
        print(line.strip())
```
